# Test 2 — Persistance Jupyter & Personnalisation (`fig.style` / `fig[x, y].style`)
Chaque cellule conserve son rendu interactif (fond dégradé/image, voile translucide adaptatif, lignes d'accent courbées et survol progressif sans *popping*).

```python
import math
import fluxio as fx

fig = fx.Figure(rows=1, cols=2, title="Expérience Thermique", size=(860, 340))
fig.style(
    background=fx.Gradient(["#0b1120", "#1e293b"], angle=125),
    accent_side="left",
    radius=16,
)
fig[0, 0].number(312.4, unit="K", precision=1, label="Température Initiale")
fig
```

```python
# Calcul intermédiaire : l'affichage de la Cellule 1 juste au-dessus reste intact !
rendement = round(0.874 * 100, 1)
print(f"Rendement calculé : {rendement} %")
```

```python
# On personnalise uniquement la case (0, 1) avec la ligne courbée à droite
fig[0, 1].style(accent_side="right", accent="#34d399").number(
    rendement, unit="%", precision=1, label="Rendement Mesuré"
)
print(fig)
```

```python
# Diapo claire en dégradé doux -> voile sombre automatique sur les cartes !
slide = fx.Figure(title="Diapo Entreprise — Charte Claire", size=(860, 320))
slide.style(
    theme="presentation",
    background=fx.Gradient(["#f8fafc", "#e0e7ff"], angle=135),
    accent="#4f46e5",
    accent_side="top",
    radius=18,
)
slide.number(
    lambda t: 120 + 25 * math.sin(t * 2),
    prefix="€",
    unit="k",
    precision=1,
    label="Flux Trésorerie",
)
slide
```
