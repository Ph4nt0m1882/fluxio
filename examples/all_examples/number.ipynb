# `Number` — Indicateur KPI Numérique (`fx.Number` / `fig.number()`)

Hiérarchie OOP : `Visual` → `Indicator` → **`Number`**.
Chaque cellule ci-dessous embarque un viewer `<canvas>` interactif autonome (survol progressif, zoom molette/glisser, capture PNG, plein écran, et contrôles Play/Pause sur les valeurs dynamiques).

## 1. Exemple minimal (1 ligne)

```python
import fluxio as fx

fx.Number(42.5, unit="°C", label="Température SoC", size=(680, 260))
```

## 2. Formatage (`prefix`, `unit`, `precision`, `description`) & Côtés d'accent (`accent_side`)

```python
fig = fx.Figure(
    rows=1,
    cols=3,
    title="KPIs Financiers & Système",
    subtitle="Bordures effilées à gauche, en haut et à droite sur fond dégradé",
    size=(940, 320),
)
fig.style(
    background=fx.Gradient(["#090d16", "#0f172a", "#1e1b4b"], angle=135),
    radius=20,
    accent_width=7.5,
)

fig[0, 0].style(accent="#38bdf8", accent_side="left").number(
    128450.8, prefix="€ ", precision=0, label="Revenu Mensuel", description="+14.2% vs M-1"
)
fig[0, 1].style(accent="#a78bfa", accent_side="top").number(
    99.982, unit="%", precision=3, label="Disponibilité SLA", description="Objectif > 99.95%"
)
fig[0, 2].style(accent="#f97316", accent_side="right").number(
    13.7, unit="ms", precision=1, label="Latence P95", description="Cluster EU-West"
)
fig
```

## 3. Les 3 variantes visuelles (`classic`, `reverse`, `text`)

```python
variants = fx.Figure(
    rows=1,
    cols=3,
    title="Comparaison des 3 Variantes",
    size=(940, 320),
)
variants.style(background=fx.Gradient(["#090d16", "#172554"], angle=125), radius=18)

variants[0, 0].style(variant="classic", accent="#38bdf8", accent_side="left").number(
    43.8, unit="°C", precision=1, label="Classic", description="Voile translucide adaptatif"
)
variants[0, 1].style(variant="reverse", accent="#059669", accent_side="top").number(
    43.8, unit="°C", precision=1, label="Reverse", description="Fond plein & encre contrastée"
)
variants[0, 2].style(variant="text", accent="#f97316", accent_side="right").number(
    43.8, unit="°C", precision=1, label="Text (TUI)", description="Matrice 6x3 type btop"
)
variants
```

## 4. Dynamique Temps Réel : Fonction `f(t)` & Générateur Python (`yield`)
Utilisez les boutons **Play / Pause** et **Replay** en haut à droite de la figure :

```python
import itertools
import math

live_num = fx.Figure(
    rows=1,
    cols=2,
    title="Flux Temps Réel (5s pré-calculées en mode Notebook)",
    size=(880, 310),
    fps=24,
)
live_num.style(background=fx.Gradient(["#090d16", "#1e1b4b"], angle=135), radius=20)

live_num[0, 0].style(accent="#f97316", accent_side="left").number(
    lambda t: 850.0 + 110.0 * math.sin(t * 2.4),
    unit="°C",
    precision=1,
    label="Chambre Magmatique — f(t)",
)
live_num[0, 1].style(variant="reverse", accent="#2563eb", accent_side="right").number(
    (i * 18 for i in itertools.count(1)),
    unit="pkt",
    precision=0,
    label="Paquets Réseau — Générateur",
)
live_num
```

## 5. Rendu Terminal Console (`print`)

```python
# Le même objet s'affiche directement en mode texte Unicode / ANSI TrueColor :
n_console = fx.Number(99.95, unit="%", precision=2, label="Availability", description="SLA > 99.9%")
print(n_console.to_text(ansi=False, width=68))
```
